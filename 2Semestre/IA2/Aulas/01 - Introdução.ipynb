## O que é aprendizado de máquina (Machine Learning)?

Literalmente ensinamos uma máquina (algoritmo) para que ela possa receber entradas (dados) e fazer previsões que podem então ser usadas no processo de tomada de decisão.

“Dados: o novo ouro”.

Com dados abundantes, o aprendizado de máquina cresceu em importância.

## Quando usar aprendizado de máquina?

Tarefas que envolvem mapeamento entre entradas e saídas bem definidas.

Quando grandes conjuntos de dados estão disponíveis ou podem ser criados contendo pares de entradas e saídas.

Tarefas que não exigem explicações detalhadas de como a decisão é tomada.

Tarefas que tenham feedback claro e objetivos bem definidos.

Tarefas que não requerem longas cadeias de raciocínio lógico.

Tarefas tolerantes a erros que não requerem soluções ideais.

O papel em estudo não muda muito rapidamente ao longo do tempo.

Exemplos:
- Detecção de Fraudes
- Sistemas de Recomendação
- Mecanismos de Busca
- Sistemas de Vigilância em Vídeo
- Reconhecimento de Manuscrito
- Processamento de Linguagem Natural
- Bots de Serviço ao Cliente
- Segurança de TI
- Manutenção Preditiva
- Previsão de Demanda
- Logística: Rotas e horários
- Negociação Financeira
- Diagnóstico de Cuidados de Saúde
- Veículos Autônomos

## Aprendizado de máquina na prática

1. **Definição dos objetivos**: trata-se do primeiro passo de qualquer análise. Devemos saber onde queremos chegar!
2. **Coleta dos dados**: envolve a coleta de dados que o algoritmo utilizará para gerar conhecimento;
3. **Exploração e preparação dos dados**: ajuste dos dados;
4. **Criação do modelo**: depois dos dados preparados, já é possível dizer o que é possível aprender deles, e como;
5. **Avaliação dos modelos**: avaliamos a qualidade do aprendizado, não
pode ser pouco (*underfitting*) nem decorar os dados (*overfitting*);
6. **Melhoria do modelo**: se necessário, podemos melhorar o desempenho do
modelo através de novas estratégias.

## Etapas de ML
1. Coleta de dados -> sensoriamento/pesquisa/experimentos
2. Visualização e preparação de dados -> matplotlib/seaborn
3. Escolha do tipo de aprendizagem (depende da aplicação, dados e
seu formato) -> supervisionado/não supervisionado/reforço
4. Criação de modelo e treinamento de algoritmo -> divisão entre treino e teste
5. Avaliação do modelo -> MAE, cross validation, accuracy
6. Atualização do modelo -> hyperparameter tuning, optimizations
7. Retreinamento e Reavaliação -> 4) e 5)

##Erros em ML

**Bias**
* Treinados em conjuntos de dados contendo dados desiguais ou incompletos.
* Modelos tendenciosos.

**Underfitting X Overfitting**

image.png

Fonte: [https://www.analyticsvidhya.com/blog/2020/02/underfitting-overfitting-best-fitting-machine-learning/](https://www.analyticsvidhya.com/blog/2020/02/underfitting-overfitting-best-fitting-machine-learning/)

**Leakage**

* Uso de informações no processo de treinamento do modelo que não se esperaria que estivessem disponíveis no momento da previsão.

* O vazamento geralmente é sutil e indireto, dificultando a detecção e eliminação.
