- **É um algoritmo muito poderoso porém pode demorar muito tempo para treinar os dados**

Problema: encontrar hiperplano que separa as classificações de forma a garantir a **maior margem**.

Sempre encontra-se a melhor solução. Portanto é um modelo **robusto**.

Chance de overfitting reduzida.

Requer todo o dataset para treinar, mas após isso a classificação é rápida: necessita apenas dos support vectors.

O número de support vectors é limitado a **metade** do tamanho do dataset. Na realidade, não chega nem perto deste valor.

#Hard-Margins SVM

Separa duas classes que são perfeitamente linearmente separáveis, w1 e w2.

Com um dataset de wi pontos, a distância de qualquer ponto até o hiperplano é |g(x)|/||w||.

image.png

O objetivo do SVM é achar w e b tais que o valor de g(x) =1 para os pontos mais próximos da classe w1 e -1 para os pontos mais próximos da classe w2.

2/||w|| = margem

#Soft-Margin SVM

Usado para dados que não são completamente linearmente separáveis. Adiciona possíveis erros, as slack variables (epsilon).

O SVM então não procura um hiperplano que vai classificar todo o seu conjunto perfeitamente. Procura um hiperplano que vai classificar a maior parte do seu conjunto corretamente. Pontos próximos a fronteira podem ser classificados incorretamente.

#Kernel SVM - The "Kernel Trick"

O que fazer quando seus dados não são linearmente separáveis, nem com uma margem de erro (Soft-Margin)?

Utilizamos um kernel que leva os pontos xi para uma dimensão maior (usualmente muito maior), em que eles poderão ser linearmente separáveis.

image.png



image.png

A seleção do kernel depende muito das especificidades dos dados. Por exemplo, o kernel linear - o mais simples de todos - é útil em grandes vetores de dados esparsos. O kernel polinomial é amplamente utilizado no processamento de imagens, enquanto o ANOVA RB
kernel é normalmente reservado para tarefas de regressão. Os RBFs Gaussianos e Laplace são kernels de uso geral que são aplicados principalmente na ausência de conhecimento prévio.

Selecionar o Kernel e os parâmetros do Kernel é **essencial** para a resposta do problema.

#Multiclasse SVM

Quando precisamos classificar nossos dados em mais de duas categorias, uma forma comum de simplificar o problema é dividir a tarefa de classificação multiclasse em várias classificações binárias. As abordagens mais usadas para isso são one-vs-one e one-vs-all (também chamada de one-vs-rest).


Imagens desta seção de:

https://www.baeldung.com/cs/svm-multiclass-classification

image.png

**One vs the rest**




Na abordagem one-vs-rest, cada classe é comparada contra todas as outras combinadas. No exemplo das classes A, B e C, as comparações seriam (A vs B+C), (B vs A+C) e (C vs A+B). A classe que superar as demais será escolhida. Utilizando os mesmos dados do exemplo anterior, as comparações resultariam em (B+C, A+C, C), e, novamente, a classe C seria a escolhida.


https://scikit-learn.org/stable/modules/generated/sklearn.multiclass.OneVsRestClassifier.html

```decision_function_shape`: {‘ovo’, ‘ovr’}, default=’ovr’```

https://scikit-learn.org/stable/modules/generated/sklearn.svm.SVC.html

image.png

**One against one**

No método one-vs-one, o problema é decomposto em classificações binárias entre todos os pares de classes. Por exemplo, se tivermos três classes A, B e C, faremos as comparações (A vs B), (A vs C) e (B vs C). A classe que receber mais votos nas comparações será a escolhida. Supondo que um SVM escolha A em (A vs B), C em (A vs C) e C em (B vs C), o resultado final será (A, C, C), e a classe C será a selecionada.

#Uso de Feature Scaling

O uso de feature scaling é essencial para alguns algoritmos de aprendizado de máquina, especialmente aqueles que consideram distâncias entre observações porque a distância entre duas observações difere para casos com e sem scaling.

A distância entre os pontos de dados afeta a margem de decisão que o SVM escolhe. Em outras palavras, treinar um SVM sobre os dados com ou sem o uso de escala leva à geração de modelos diferentes.

#Aplicação em Código

````python
import numpy as np
#import matplotlib.pyplot as plt
#from matplotlib.colors import ListedColormap
import pandas as pd
import seaborn as sns

#from sklearn import tree

#from sklearn.impute import SimpleImputer
#from sklearn.compose import ColumnTransformer
#from sklearn.preprocessing import OneHotEncoder
#from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

#from sklearn.linear_model import LinearRegression
#from sklearn.tree import DecisionTreeRegressor
#from sklearn.ensemble import RandomForestRegressor

#from sklearn.tree import DecisionTreeClassifier
#from sklearn.ensemble import RandomForestClassifier

from sklearn.svm import SVC

#from sklearn.model_selection import cross_val_score
#from sklearn.model_selection import GridSearchCV

#from sklearn.metrics import r2_score
#from sklearn.metrics import mean_absolute_error

from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import ConfusionMatrixDisplay
````

````python
titanic = sns.load_dataset("titanic")
````

Retirando a coluna deck por ter muitos NaN.

````python
titanic.drop('deck',  axis=1, inplace=True)
````

Retirando as linhas com NaN.

````python
dataset2 = titanic.dropna(axis=0) #retira as linhas com NaN
````

````python
y = dataset2.survived
````

Primeiro teste com apenas os dados numéricos.

````python
features = ['pclass', 'age', 'sibsp', 'fare', 'parch']

X = dataset2[features]
````

````python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
````

````python
from sklearn.svm import SVC
classifier = SVC(kernel = 'linear', random_state = 0)
classifier.fit(X_train, y_train)
````

````python
y_pred = classifier.predict(X_test)
````

````python
accuracy_score(y_test, y_pred)
````

````python

classifier2 = SVC(kernel = 'rbf', random_state = 0)
classifier2.fit(X_train, y_train)
````

````python
y_pred2 = classifier2.predict(X_test)
````

````python
accuracy_score(y_test, y_pred2)
````

Acurácia ruim, abaixo do que obtido com outros modelos.

Vamos tentar com as melhores features obtidas para o random forest: adicionando 'adult_male'.

````python
features = ['pclass', 'age', 'sibsp', 'fare','adult_male']

X = dataset2[features].values
y = dataset2.survived
````

Codificando a variável categórica:

````python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

X = np.array(ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [-1])], remainder='passthrough').fit_transform(X))
````

````python
X
````

````python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
````

Testando vários kernels disponíveis no sklearn:

````python
classifier3 = SVC(kernel = 'linear', random_state = 0)
classifier3.fit(X_train, y_train)
classifier4 = SVC(kernel = 'rbf', random_state = 0)
classifier4.fit(X_train, y_train)
classifier5 = SVC(kernel = 'sigmoid', random_state = 0)
classifier5.fit(X_train, y_train)
classifier6 = SVC(kernel = 'poly', degree = 3, random_state = 0)
classifier6.fit(X_train, y_train)
````

````python
y_pred3 = classifier3.predict(X_test)
y_pred4 = classifier4.predict(X_test)
y_pred5 = classifier5.predict(X_test)
y_pred6 = classifier6.predict(X_test)
````

````python
accuracy_score(y_test, y_pred3)
````

````python
cm = confusion_matrix(y_test, y_pred3)

disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot();
````

````python
accuracy_score(y_test, y_pred4)
````

````python
accuracy_score(y_test, y_pred5)
````

````python
accuracy_score(y_test, y_pred6)
````

#Ponto sobre acurácia e seu paradoxo

Por que precisamos também ver a matriz de confusão? Quando temos um dataset muito desbalanceado, podemos criar modelos "burros":

````python
cm1 = np.array([[(9500,150),(50,100)]])
print(cm1)
acuracia = (9500+100)/(9500+150+50+100)
print(acuracia)
````

````python
cm2 = np.array([[(9500+150,0),(50+100,0)]])
print(cm2)
acuracia = (9500+150)/(9500+150+50+100)
print(acuracia)
````
