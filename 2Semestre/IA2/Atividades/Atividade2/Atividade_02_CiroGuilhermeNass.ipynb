### Dataset escolhido: **penguins**

Dados de 344 pinguins de 3 especies (Adelie, Chinstrap, Gentoo) coletados em 3 ilhas do arquipelago Palmer (Antartica), com medidas de bico, nadadeira, massa corporal e sexo.

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

df = sns.load_dataset("penguins")
df = df.dropna()
df.head()
```

```python
df.info()
df.describe()
```

---
### Grafico 1 - Histograma da massa corporal por especie

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="body_mass_g", hue="species", kde=True, bins=25)
plt.title("Distribuicao da massa corporal por especie")
plt.xlabel("Massa corporal (g)")
plt.ylabel("Frequencia")
plt.show()
```

**Por que este grafico:** o histograma e a forma padrao de visualizar a *distribuicao* de uma variavel numerica; separando por cor (`hue`) comparo as tres especies na mesma escala.

**O que obtive:** a distribuicao geral e bimodal. Adelie e Chinstrap se sobrepoem quase totalmente (~3.000-4.500 g), enquanto Gentoo forma um grupo claramente mais pesado (~4.500-6.000 g). Ou seja, a massa corporal sozinha ja separa Gentoo das outras duas, mas nao distingue Adelie de Chinstrap.

---
### Grafico 2 - Dispersao: comprimento x profundidade do bico

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="bill_length_mm", y="bill_depth_mm",
                hue="species", style="species", s=70)
plt.title("Comprimento x profundidade do bico por especie")
plt.xlabel("Comprimento do bico (mm)")
plt.ylabel("Profundidade do bico (mm)")
plt.show()
```

**Por que este grafico:** o scatterplot mostra a *relacao entre duas variaveis numericas* e revela agrupamentos naturais.

**O que obtive:** os tres grupos ficam quase perfeitamente separados. Adelie tem bico curto e profundo, Chinstrap bico longo e profundo, Gentoo bico longo e raso. Este par de atributos e o melhor candidato para classificar a especie, pois resolve exatamente a ambiguidade Adelie/Chinstrap que o Grafico 1 deixou em aberto.

Observacao: no dataset inteiro (sem separar por especie) a correlacao entre as duas medidas e negativa, mas dentro de cada especie e positiva - um caso classico de **Paradoxo de Simpson**.

---
### Grafico 3 - Boxplot do comprimento da nadadeira por especie e sexo

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="species", y="flipper_length_mm", hue="sex")
plt.title("Comprimento da nadadeira por especie e sexo")
plt.xlabel("Especie")
plt.ylabel("Comprimento da nadadeira (mm)")
plt.show()
```

**Por que este grafico:** o boxplot resume mediana, quartis, amplitude e outliers, sendo ideal para *comparar grupos categoricos*; o `hue="sex"` acrescenta uma segunda dimensao de comparacao.

**O que obtive:** Gentoo tem nadadeiras bem maiores (mediana ~216 mm) que Adelie e Chinstrap (~190-196 mm). Em todas as especies os machos sao maiores que as femeas, com deslocamento consistente das medianas - ha **dimorfismo sexual**. As caixas sao compactas e com poucos outliers, indicando dados limpos e baixa variabilidade interna.

---
### Grafico 4 - Mapa de calor das correlacoes

```python
num = df.select_dtypes(include="number")

plt.figure(figsize=(7, 5))
sns.heatmap(num.corr(), annot=True, cmap="coolwarm", vmin=-1, vmax=1, fmt=".2f")
plt.title("Correlacao entre as variaveis numericas")
plt.show()
```

**Por que este grafico:** o heatmap mostra de uma so vez a *forca e o sinal de todas as correlacoes* entre as variaveis numericas, algo dificil de enxergar rapidamente em uma tabela.

**O que obtive:** correlacao forte e positiva entre `flipper_length_mm` e `body_mass_g` (~0,87): pinguins maiores sao mais pesados. `bill_depth_mm` tem correlacao negativa com massa e nadadeira (~-0,47 e -0,58), efeito puxado pelos Gentoo (grandes e de bico raso). Como nadadeira e massa carregam quase a mesma informacao, ha redundancia: para um modelo preditivo, uma das duas poderia ser descartada.

---
### Grafico 5 - Contagem de pinguins por ilha e especie

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="island", hue="species")
plt.title("Quantidade de pinguins por ilha e especie")
plt.xlabel("Ilha")
plt.ylabel("Quantidade")
plt.show()
```

**Por que este grafico:** o countplot e a escolha natural para *frequencia de variaveis categoricas*, permitindo ver a composicao de cada ilha.

**O que obtive:** a distribuicao geografica nao e uniforme. Biscoe abriga apenas Gentoo e Adelie, Dream apenas Chinstrap e Adelie, e Torgersen **exclusivamente** Adelie. A ilha e, portanto, por si so um forte indicador da especie: quem esta em Torgersen e Adelie, e Gentoo e Chinstrap nunca aparecem juntas. Adelie e a unica especie presente nas tres ilhas.

---
### Conclusao

As cinco visualizacoes se complementam: o **histograma** mostrou a distribuicao da massa e separou os Gentoo; o **scatterplot** das medidas do bico separou as tres especies; o **boxplot** comparou grupos e revelou o dimorfismo sexual; o **heatmap** quantificou as relacoes entre as variaveis e apontou redundancia; e o **countplot** descreveu a distribuicao geografica.

Para classificar a especie de um pinguim, os atributos mais informativos sao **comprimento e profundidade do bico**, complementados pela **ilha** e pelo **comprimento da nadadeira**.
