### Previsão da Expectativa de Vida com Regressão Linear

Nesta atividade, você utilizará regressão linear para prever a expectativa de vida com base no Índice de Massa Corporal (IMC). Você trabalhará com dados reais do Gapminder, que contêm a expectativa média de vida ao nascer e o IMC médio de mulheres ao redor do mundo.

#### Objetivo:
Criar um modelo de regressão linear que estime a expectativa de vida com base no IMC médio de diferentes países.

#### Descrição dos Dados:
Você encontrará o arquivo de dados no github: `https://raw.githubusercontent.com/OrlandoIFPR/ml2/refs/heads/main/Dataframes/Atividades/04-bmi_and_life_expectancy.csv`. Este arquivo contém três colunas:

- País (Country): Nome do país.
- Expectativa de Vida (Life expectancy): Expectativa média de vida ao nascer em anos.
- IMC (BMI): Índice de Massa Corporal médio naquele país.

#### Etapas da Atividade:

1. Faça uma cópia de arquivo para o seu google drive, renomeie o arquivo com o seguinte nome: "Atividade 04 - Seu Nome.ipynb", em que "Seu Nome" deve substituido pelo seu nome.
2. Carregue os Dados `04-bmi_and_life_expectancy.csv`.
4. Análise os Dados carregados.
5. Crie o Modelo de Regressão Linear, pode utilizar todos os dados para treinar o modelo e não precisa avaliar o modelo.
6. Faça a previsão de 5 diferentes IMC.
